

## Useful `read_csv` / `read_excel` parameters

These come up constantly with messy real-world exports:

| Parameter | What it does |
|---|---|
| `usecols=['A','B']` | Load only specific columns (saves memory) |
| `dtype={'Vendor Code': str}` | Force a column's data type on load (avoids e.g. leading zeros being dropped) |
| `parse_dates=['Invoice Date']` | Automatically convert a column to datetime while reading |
| `na_values=['NA', 'N/A', '-']` | Tell pandas which strings should be treated as missing |
| `nrows=100` | Load only the first N rows — great for quickly previewing a huge file |


````python
# Example: load with dtype control and parsed dates
df = pd.read_csv(
    "../data/synthetic_ap_invoices.csv",
    dtype={"Vendor Code": str, "PO Number": str},
    parse_dates=["Invoice Date"],
)
df.dtypes
````
