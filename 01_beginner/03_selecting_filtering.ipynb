# 03 · Selecting & Filtering Data

This is the pandas equivalent of a SQL `WHERE` clause and `SELECT` column list — arguably the
skill you'll use most, every single day.


```python
import pandas as pd

df = pd.read_csv("../data/synthetic_ap_invoices.csv", parse_dates=["Invoice Date"])
df.head(2)
```

## Selecting columns

```python
df['Invoice Amount']          # a single column -> returns a Series
```

```python
df[['Invoice Number', 'Invoice Amount', 'Hold Reason']]   # multiple columns -> returns a DataFrame
```

## `.loc` vs `.iloc`

- `.loc` selects by **label** (column name, index label)
- `.iloc` selects by **integer position** (like list indexing)

This distinction trips up almost every beginner, so let's be explicit.


```python
df.loc[0]                     # row with index label 0
```

```python
df.loc[0, 'Invoice Amount']   # value at row 0, column 'Invoice Amount'
```

```python
df.iloc[0]                    # first row by position (same result here since index is default 0..n)
```

```python
df.iloc[0:5, 0:3]             # first 5 rows, first 3 columns, by position
```

## Boolean filtering — the real workhorse

You write a condition, pandas returns True/False for every row, then you use that to filter.


```python
on_hold = df[df['Hold Reason'].notna()]
on_hold.shape
```

```python
high_value = df[df['Invoice Amount'] > 10000]
high_value[['Invoice Number', 'Invoice Amount']].head()
```

```python
# Combine conditions: & = AND, | = OR — always wrap each condition in parentheses
de_high_value_holds = df[
    (df['Region'] == 'DE10') &
    (df['Invoice Amount'] > 5000) &
    (df['Hold Reason'].notna())
]
de_high_value_holds.shape
```

```python
# .isin() for matching against a list of values — cleaner than chained OR conditions
priority_reasons = df[df['Hold Reason'].isin(['Price Difference', 'Duplicate Invoice'])]
priority_reasons['Hold Reason'].value_counts()
```

## `.query()` — a readable alternative for complex filters

Handy when conditions get long; reads closer to plain English / SQL.


```python
df.query("Region == 'UK30' and `Invoice Amount` > 8000").head()
```
