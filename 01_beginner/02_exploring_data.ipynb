# 02 · Exploring a Dataset

Before you clean or analyze anything, you need to understand what you're looking at.
This is the "first 5 minutes" checklist every analyst runs on a new dataset.


```python
import pandas as pd

df = pd.read_csv("../data/synthetic_ap_invoices.csv", parse_dates=["Invoice Date"])
df.shape  # (rows, columns) — always check this first
```

```python
df.head()   # first 5 rows
```

```python
df.tail(3)  # last 3 rows
```

```python
df.info()   # column names, non-null counts, dtypes — THE most useful command for a new dataset
```

```python
df.describe()  # summary stats for numeric columns: count, mean, std, min, max, quartiles
```

## Checking individual columns

`.info()` and `.describe()` give you the big picture. Now zoom into specific columns.

```python
df.columns.tolist()  # list of all column names
```

```python
df['Hold Reason'].unique()  # what distinct values exist in this column?
```

```python
df['Hold Reason'].value_counts()  # count of each value (NaN excluded by default)
```

```python
df['Hold Reason'].value_counts(dropna=False)  # include the NaN count too
```

```python
df.isna().sum()  # how many missing values per column — critical for a cleaning plan
```
